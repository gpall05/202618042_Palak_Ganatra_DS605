# Task 2: Vectorized Arithmetic and Linear Algebra

```python
# Create two matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)


# Matrix addition
addition = A + B
print("\nA + B:")
print(addition)


# Element-wise multiplication
elementwise_multiplication = A * B
print("\nElement-wise multiplication (A * B):")
print(elementwise_multiplication)


# Matrix multiplication
matrix_multiplication = A @ B
print("\nMatrix multiplication (A @ B):")
print(matrix_multiplication)


# Transpose
transpose_A = A.T
print("\nTranspose of A:")
print(transpose_A)


# Determinant
determinant_A = np.linalg.det(A)
print("\nDeterminant of A:", determinant_A)


# Inverse
inverse_A = np.linalg.inv(A)
print("\nInverse of A:")
print(inverse_A)


# Verify A @ A^-1 = I
verification = A @ inverse_A

print("\nA @ A^-1:")
print(verification)

print("\nVerification using np.allclose():")
print(np.allclose(verification, np.eye(2)))
```

# ## Task 5: Filtering and Querying

```python
# 1. Male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("1. Male passengers older than 50:", len(male_over_50))


# 2. Female first-class passengers and survival percentage
female_first_class = df[
    (df["Sex"] == "female") & (df["Pclass"] == 1)
]

female_first_class_count = len(female_first_class)
female_first_class_survival = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", female_first_class_count)
print(
    "   Survival percentage:",
    round(female_first_class_survival, 2),
    "%"
)


# 3. Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()

condition_3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print(
    "\n3. Age 20-40, Fare above median, and survived:",
    len(condition_3)
)


# 4. Travelling alone, age < 30, and did not survive
condition_4 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print(
    "\n4. Travelling alone, age < 30, and did not survive:",
    len(condition_4)
)


# 5. Embarked = S, Pclass 2 or 3,
#    Fare above Southampton median
southampton = df[df["Embarked"] == "S"]
southampton_fare_median = southampton["Fare"].median()

condition_5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_fare_median)
]

print(
    "\n5. Embarked S, Pclass 2 or 3, "
    "Fare above Southampton median:",
    len(condition_5)
)
```

# task 6: groupby()

```python
# 1. Survival rate by Sex
print("1. Survival Rate by Sex:")
display(df.groupby("Sex")["Survived"].mean())


# 2. Survival rate by Pclass
print("\n2. Survival Rate by Pclass:")
display(df.groupby("Pclass")["Survived"].mean())


# 3. Average Age and Fare by Pclass
print("\n3. Average Age and Fare by Pclass:")
display(
    df.groupby("Pclass")[["Age", "Fare"]].mean()
)


# 4. Passenger count and survival rate by Sex and Pclass
print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")

sex_pclass = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

display(sex_pclass)


# 5. Passenger count, average Fare, and survival rate by Embarked
print("\n5. Passenger Count, Average Fare, and Survival Rate by Embarked:")

embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

display(embarked_stats)
```

# task 7 Missing Values & Fare Outliers

```python
# Missing value count
missing_count = df.isnull().sum()

# Missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing Values Summary:")
display(missing_summary)


# Plot missing values
plt.figure(figsize=(10, 5))
missing_count[missing_count > 0].plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Missing Age before filling
print("Missing Age before filling:", df["Age"].isnull().sum())

# Fill missing Age with mean
age_mean = df["Age"].mean()
df["Age"] = df["Age"].fillna(age_mean)

print("Missing Age after filling:", df["Age"].isnull().sum())
```

```python
# Demonstration of different Age imputation methods

age_original = df["Age"].copy()

# Mean
age_mean_imputed = age_original.fillna(age_original.mean())

# Median
age_median_imputed = age_original.fillna(age_original.median())

# Mode
age_mode_imputed = age_original.fillna(age_original.mode()[0])

# Random value
np.random.seed(42)
age_random_imputed = age_original.copy()

missing_indices = age_random_imputed[age_random_imputed.isna()].index

random_values = np.random.choice(
    age_original.dropna(),
    size=len(missing_indices),
    replace=True
)

age_random_imputed.loc[missing_indices] = random_values

print("Mean imputation completed.")
print("Median imputation completed.")
print("Mode imputation completed.")
print("Random-value imputation completed.")
```

```python
# Calculate Q1 and Q3
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Fare Q1:", Q1)
print("Fare Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(fare_outliers))
```

# Task 8 Features & Pivot Table

```python
# Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Create IsAlone
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("First 10 rows with new features:")
display(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))


# Create pivot table
pivot_table = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("\nSurvival Rate by Sex and Pclass:")
display(pivot_table)


# Find highest and lowest survival groups
highest_group = pivot_table.stack().idxmax()
lowest_group = pivot_table.stack().idxmin()

highest_rate = pivot_table.stack().max()
lowest_rate = pivot_table.stack().min()

print("Highest survival group:", highest_group)
print("Highest survival rate:", round(highest_rate * 100, 2), "%")

print("\nLowest survival group:", lowest_group)
print("Lowest survival rate:", round(lowest_rate * 100, 2), "%")
```

# Task 9: Visualizations and Observations

```python
# Correlation heatmap

numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Survival rate by Sex

survival_by_sex = df.groupby("Sex")["Survived"].mean()

plt.figure(figsize=(6, 5))
survival_by_sex.plot(kind="bar")
plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("survival_rate_by_sex.png", dpi=300, bbox_inches="tight")
plt.show()

print("Survival rate by Sex:")
print((survival_by_sex * 100).round(2))
```

```python
# Age vs Fare, distinguished by survival

plt.figure(figsize=(8, 6))

for survival_status in [0, 1]:
    subset = df[df["Survived"] == survival_status]
    
    plt.scatter(
        subset["Age"],
        subset["Fare"],
        label=f"Survived = {survival_status}",
        alpha=0.5
    )

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend()
plt.tight_layout()
plt.savefig("age_vs_fare.png", dpi=300, bbox_inches="tight")
plt.show()
```
